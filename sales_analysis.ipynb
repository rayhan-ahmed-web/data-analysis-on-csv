# Sales Data Analysis using Pandas

This notebook analyzes a sample sales dataset using Pandas and Matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Load the CSV file

```python
sales = pd.read_csv('sales_data.csv')
sales['Date'] = pd.to_datetime(sales['Date'])
sales.head()
```

## 2. Explore the dataset

```python
print('Shape:', sales.shape)
print('\nColumns:', list(sales.columns))
print('\nData types:')
print(sales.dtypes)
print('\nMissing values:')
print(sales.isnull().sum())
```

## 3. Overall sales metrics

```python
total_sales = sales['Total_Sales'].sum()
total_quantity = sales['Quantity'].sum()
average_order = sales['Total_Sales'].mean()
print(f'Total sales: ₹{total_sales:,.0f}')
print(f'Total quantity sold: {total_quantity}')
print(f'Average order value: ₹{average_order:,.2f}')
```

## 4. Sales by category

```python
category_sales = sales.groupby('Category')['Total_Sales'].sum().sort_values(ascending=False)
category_sales
```

```python
category_sales.plot(kind='bar', title='Total Sales by Category', xlabel='Category', ylabel='Sales (₹)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 5. Sales by region

```python
region_sales = sales.groupby('Region')['Total_Sales'].sum().sort_values(ascending=False)
region_sales
```

```python
region_sales.plot(kind='bar', title='Total Sales by Region', xlabel='Region', ylabel='Sales (₹)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 6. Product performance

```python
product_sales = sales.groupby('Product').agg(Total_Sales=('Total_Sales','sum'), Quantity_Sold=('Quantity','sum')).sort_values('Total_Sales', ascending=False)
product_sales
```

## 7. Monthly sales trend

```python
monthly_sales = sales.groupby(sales['Date'].dt.to_period('M'))['Total_Sales'].sum()
monthly_sales
```

```python
monthly_sales.astype(float).plot(kind='line', marker='o', title='Monthly Sales Trend', xlabel='Month', ylabel='Sales (₹)')
plt.tight_layout()
plt.show()
```

## 8. Filtering and highest-value order

```python
high_value_orders = sales[sales['Total_Sales'] >= 100000]
print('Orders with sales >= ₹100,000:')
display(high_value_orders)

best_order = sales.loc[sales['Total_Sales'].idxmax()]
print('Highest-value order:')
display(best_order)
```

## 9. Key findings

- Total revenue and quantity are calculated with Pandas aggregation functions.
- Category and regional performance are compared using `groupby()` and `sum()`.
- Product performance is ranked by total revenue.
- Monthly revenue is visualized to identify sales trends.
- Filtering identifies high-value orders for further analysis.
